```python
start_scope() 
defaultclock.dt = 0.01*ms

# === 神经元参数（同上） ===
Cm  = 1*uF/cm**2
El  = 10.613*mV
ENa = 115*mV
EK  = -12*mV
gl  = 0.3*msiemens/cm**2
gNa = 120*msiemens/cm**2
gK  = 36*msiemens/cm**2

E_exc = 0*mV      # 兴奋性反转电位（类 AMPA）
E_inh = -80*mV    # 抑制性反转电位（类 GABA）
tau_e = 5*ms
tau_i = 10*ms

eqs_HH_syn = '''
dv/dt = (gl*(El-v) + gNa*m**3*h*(ENa-v) + gK*n**4*(EK-v)
         + ge*(E_exc-v) + gi*(E_inh-v)
         + I_inj/area) / Cm : volt

dge/dt = -ge/tau_e : siemens
dgi/dt = -gi/tau_i : siemens

I_inj : amp
area  : metre**2

dm/dt = alpham*(1-m) - betam*m : 1
dn/dt = alphan*(1-n) - betan*n : 1
dh/dt = alphah*(1-h) - betah*h : 1

alpham = (0.1/mV)*10*mV/exprel((-v+25*mV)/(10*mV))/ms : Hz
betam  = 4*exp(-v/(18*mV))/ms : Hz
alphah = 0.07*exp(-v/(20*mV))/ms : Hz
betah  = 1/(exp((-v+30*mV)/(10*mV)) + 1)/ms : Hz
alphan = (0.01/mV)*10*mV/exprel((-v+10*mV)/(10*mV))/ms : Hz
betan  = 0.125*exp(-v/(80*mV))/ms : Hz
'''
```
